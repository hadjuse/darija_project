```python
import torch
from torch.utils.data import Dataset

class AudioDataset(Dataset):

    def __init__(self, wav_dir, mel_dir):

        self.wav_files = sorted(os.listdir(wav_dir))
        self.mel_files = sorted(os.listdir(mel_dir))

        self.wav_dir = wav_dir
        self.mel_dir = mel_dir

    def __len__(self):
        return len(self.wav_files)

    def __getitem__(self, idx):

        wav = np.load(f"{self.wav_dir}/{self.wav_files[idx]}")
        mel = np.load(f"{self.mel_dir}/{self.mel_files[idx]}")

        wav = torch.tensor(wav).float()
        mel = torch.tensor(mel).float()

        return mel, wav
```
